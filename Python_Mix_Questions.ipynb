### G. Concatenation patterns: join, f-strings, format (5)

- Given cols = ["age","income","zip"], produce a SQL select clause using ", ".join(cols).

```python
cols = ["age","income","zip"] 
sql1 = "SELECT "+",".join(cols) +" FROM table;" 
print (sql1)
```

```python
sql2 = f"SELECT {",".join(cols)} FROM table;"
print (sql2)
```

```python
sql3 = "SELECT {} FROM table;".format(", ".join(cols))
print (sql3)
```

- Use aligned formatting to print a table header 'Name' left-justified in 20 chars.

```python
print (f"{"Name":<20}{"Age":<10}{"City":<15}")
print (f"{"Darshit":<20}{"24":<10}{"Rajkot":<15}")
```

```python
print ("{:<20}{:<10}{:<15}".format("Name", "Age", "City"))
print ("{:<20}{:<10}{:<15}".format("Darshit","24", "Rajkot"))
```

```python
print("Name".ljust(20) + "Age".ljust(10) + "City".ljust(15))
print("Darshit".ljust(20) + "24".ljust(10) + "Rajkot".ljust(15))
```
